```python
total_price=sum(cafe_menu.values())
print(f"모든 메뉴 가격의 총합: {total_price}원")
```

```python
total_price = 0
ordered_items = []
print("메뉴 이름을 정확히 입력해 주세요. (주문을 마치려면 '결제'를 입력하세요)\n")
while True:
  order = input("주문할 메뉴를 입력하세요: ").strip()
  if order == '결제':
    break
  if order in cafe_menu:
    price = cafe_menu[order]
    total_price += price
    ordered_items.append(order)
    print(f" [{order}] 추가되었습니다. (현재 누적 금액: {total_price}원)")
  else:
    print('메뉴판에 없는 메뉴입니다')
if len(ordered_items) == 0:
    print("주문하신 메뉴가 없습니다.")
else:
    print(f" 주문 내역: {', '.join(ordered_items)}")
    print(f" 총 결제 금액: {total_price}원")
```
